```python
df["issue_date"] = pd.to_datetime(
    df["issue_date"],
    errors="coerce"
)
```

```python
df["issue_year"] = df["issue_date"].dt.year
df["issue_month"] = df["issue_date"].dt.month
df["issue_month_name"] = df["issue_date"].dt.month_name()
```

**Create important analytical columns**

```python
df["loan_to_income"] = (
    df["loan_amount"] / df["annual_income"]
)
```

```python
df["loan_to_income_pct"] = (
    df["loan_to_income"] * 100
)
```

```python
df["net_payment"] = (
    df["total_payment"] - df["funded_amount"]
)
```

```python
df["default_flag"] = np.where(
    df["loan_status"].isin(["Charged Off", "Default"]),
    1,
    0
)
```

**Start with basic KPIs**

```python
total_loans = df["loan_id"].nunique()

print("Total Loans:", total_loans)
```

```python
total_loan_amount = df["loan_amount"].sum()

print("Total Loan Amount:", total_loan_amount)
```

```python
total_funded = df["funded_amount"].sum()

print("Total Funded Amount:", total_funded)
```

```python
avg_interest = df["interest_rate"].mean()

print("Average Interest Rate:", round(avg_interest, 2))
```

```python
total_interest = df["total_interest"].sum()

print("Total Interest:", total_interest)
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="loan_status",
    order=df["loan_status"].value_counts().index
)

plt.title("Loan Status Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Number of Loans")
plt.xticks(rotation=45)

plt.show()
```

**Visualization 2 — Loan Grade**

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="grade",
    order=sorted(df["grade"].dropna().unique())
)

plt.title("Loan Distribution by Grade")
plt.xlabel("Loan Grade")
plt.ylabel("Number of Loans")

plt.show()
```

**Visualization 3 — Interest Rate by Grade**

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="grade",
    y="interest_rate"
)

plt.title("Interest Rate Distribution by Loan Grade")
plt.xlabel("Loan Grade")
plt.ylabel("Interest Rate (%)")

plt.show()
```

**Visualization 4 — Loan Amount Distribution**

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["loan_amount"],
    bins=30,
    kde=True
)

plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Number of Loans")

plt.show()
```

**Visualization 5 — Annual Income**
